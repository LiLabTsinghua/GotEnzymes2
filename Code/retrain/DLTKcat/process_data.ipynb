```python
import pandas as pd
import numpy as np

km = 'KCAT'
df = pd.read_csv('../../data/EITLEM_KCAT_T.csv')

# 删除Temperature列中包含NaN的行
df.dropna(subset=['Temperature'], inplace=True)

# 重命名列
df.rename(columns={'Value': 'kcat', 'Sequence': 'seq', 'Smiles': 'smiles'}, inplace=True)

# 计算温度相关的列
Temp_K = [float(T) + 273.15 for T in df['Temperature']]
Inv_Temp = [1 / (float(T) + 273.15) for T in df['Temperature']]

# 归一化函数
def scale_minmax(array, x_min, x_max):
    '''
    Normalize the feature as (x - x_min) / (x_max - x_min).
    '''
    scaled_array = [(x - x_min) / (x_max - x_min) for x in array]
    return scaled_array

# 计算归一化后的温度
T_K_norm = scale_minmax(Temp_K, min(Temp_K), max(Temp_K))
inv_T_norm = scale_minmax(Inv_Temp, min(Inv_Temp), max(Inv_Temp))

# 将归一化后的温度添加到DataFrame
df['Temp_K'] = Temp_K
df['Inv_Temp'] = Inv_Temp
df['Temp_K_norm'] = T_K_norm
df['Inv_Temp_norm'] = inv_T_norm

# 保存结果到CSV文件（可选）
df.to_csv('./data/EITLEM_{}_all.csv'.format(km), index=False)

df
```

```python
df.drop(columns=['Test'], inplace=True)
from sklearn.model_selection import train_test_split
# train_df = df[df['Test'] == 0]
# test_df = df[df['Test'] == 1]
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42)
print(train_df.shape, test_df.shape)
train_df
```

```python
train_df['Test'] = 0
test_df['Test'] = 1
train_df.to_csv('./data/EITLEM_{}_train.csv'.format(km), index=False)
test_df.to_csv('./data/EITLEM_{}_test.csv'.format(km), index=False)
```

```python
all_df = pd.concat([train_df, test_df])
all_df.to_csv('./data/EITLEM_{}_all_new_split.csv'.format(km), index=False)
```
