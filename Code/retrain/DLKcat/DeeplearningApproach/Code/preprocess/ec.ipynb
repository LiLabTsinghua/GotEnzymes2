```python
LIST1,LIST2,LIST3,LIST4,LIST5,LIST6,LIST7=[],[],[],[],[],[],[]
def eclist():
    with open('../../Data/EC_enzyme/enzyme.dat', 'r') as rf:
        lines = rf.readlines()

    ec_list = list()
    for i in range(len(lines)):
        if lines[i].startswith('ID'):
            if lines[i + 1].startswith('DE   Transferred entry:'):
                continue
            else:
                ec = lines[i].strip().split('  ')[1].strip()
                ec_list.append(ec)
                if ec.startswith('1.'):
                    LIST1.append(ec)
                elif ec.startswith('2.'):
                    LIST2.append(ec)
                elif ec.startswith('3.'):
                    LIST3.append(ec)
                elif ec.startswith('4.'):
                    LIST4.append(ec)
                elif ec.startswith('5.'):
                    LIST5.append(ec)
                elif ec.startswith('6.'):
                    LIST6.append(ec)
                elif ec.startswith('7.'):
                    LIST7.append(ec)
    return ec_list

# 调用函数并打印结果
ec_numbers = eclist()
print(ec_numbers)
print(len(ec_numbers))
print(len(LIST1),len(LIST2),len(LIST3),len(LIST4),len(LIST5),len(LIST6),len(LIST7))
```

```python
import matplotlib.pyplot as plt

# 列表长度数据
list_lengths = [len(LIST1), len(LIST2), len(LIST3), len(LIST4), len(LIST5), len(LIST6), len(LIST7)]

# 列表名称
labels = ['ec1', 'ec2', 'ec3', 'ec4', 'ec5', 'ec6', 'ec7']

# 创建直方图
plt.figure(figsize=(10, 6))
plt.bar(labels, list_lengths, color='green')
plt.xlabel('EC Number')
plt.ylabel('Length')
plt.title('Histogram of List Lengths')
plt.show()
```

```python
import matplotlib.pyplot as plt
import json

# 读取数据
with open('../../Data/database/Kcat_combination_0918_wildtype_mutant.json', 'r') as rf:
    datasets = json.load(rf)

# 提取EC Number
ECNumber = [data['ECNumber'] for data in datasets]

# 计算每组EC Number的数量
ec1_count = len([ec for ec in set(ECNumber) if ec.startswith('1')])
ec2_count = len([ec for ec in set(ECNumber) if ec.startswith('2')])
ec3_count = len([ec for ec in set(ECNumber) if ec.startswith('3')])
ec4_count = len([ec for ec in set(ECNumber) if ec.startswith('4')])
ec5_count = len([ec for ec in set(ECNumber) if ec.startswith('5')])
ec6_count = len([ec for ec in set(ECNumber) if ec.startswith('6')])
ec7_count = len([ec for ec in set(ECNumber) if ec.startswith('7')])
ec_counts = [ec1_count, ec2_count, ec3_count, ec4_count, ec5_count, ec6_count, ec7_count]
# 列表名称
labels = ['EC1', 'EC2', 'EC3', 'EC4', 'EC5', 'EC6', 'EC7']

# 创建直方图
plt.figure(figsize=(10, 6))
plt.bar(labels, [ec1_count, ec2_count, ec3_count, ec4_count, ec5_count, ec6_count, ec7_count], color='green')
plt.xlabel('EC Number')
plt.ylabel('Number of EC Numbers')
plt.title('Histogram of EC Numbers')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
# 绘制第一个直方图
plt.bar(labels, list_lengths, color='pink', label='all EC')

# 在同一张图上绘制第二个直方图
plt.bar(labels, ec_counts, color='blue', label='our EC')
plt.xlabel('EC Number')
plt.ylabel('Counts')
plt.title('Histogram of Different EC Numbers')
plt.legend()

# 展示图形
plt.show()
```
