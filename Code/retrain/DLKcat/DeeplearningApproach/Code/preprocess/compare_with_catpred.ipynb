```python
from rdkit import Chem
import pandas as pd
import numpy as np
def smiles2inchikey0(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles)
        inchikey = Chem.MolToInchiKey(mol)
        if inchikey:
            return inchikey.split('-')[0]
        else:
            return smiles 
    except:
        return smiles

def compare_smiles_inchikey(smile1,smile2):
    inchikey1 = smiles2inchikey0(smile1)
    inchikey2 = smiles2inchikey0(smile2)
    return inchikey1 == inchikey2

# Function to split SMILES strings and flatten the list
def split_and_flatten(smiles_list):
    # Convert each entry to string, split by '.', then flatten the resulting lists into one list
    flattened_list = [item for sublist in smiles_list for item in str(sublist).split('.')]
    return np.unique(flattened_list)  # Apply unique after splitting and flattening
if __name__ == '__main__':
    df_catpred = pd.read_csv('kcat-random_trainvaltest.csv')
    df_ours = pd.read_json('../../Data/database/Kcat_combination_0918_wildtype_mutant.json')
    list_catpred_all = split_and_flatten(df_catpred['reactant_smiles'])
    list_catpred_inch = [smiles2inchikey0(smile) for smile in list_catpred_all]
    list_ours = split_and_flatten(df_ours['Smiles'])
    list_ours_inch = [smiles2inchikey0(smile) for smile in list_ours]

    set_A = set(list_catpred_inch)
    set_B = set(list_ours_inch)
    intersection = set_A & set_B
    A_not_B = set_A - set_B
    B_not_A = set_B - set_A

    intersection_list = list(intersection)
    A_not_B_list = list(A_not_B)
    B_not_A_list = list(B_not_A)

    # 打印结果
    print("交集:", len(intersection_list),intersection_list)
    print("catpred有ours无:", len(A_not_B_list),A_not_B_list)
    print("catpred无ours有:", len(B_not_A_list),B_not_A_list)
    print('catpred inch',len(set_A),'ours inch',len(set_B))
```
